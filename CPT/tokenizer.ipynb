```
from transformers import AutoTokenizer
from transformers import AutoConfig

model_id = "mistralai/Mistral-7B-v0.1"

config = AutoConfig.from_pretrained("mistralai/Mistral-7B-v0.1")
print(config.max_position_embeddings)

tokenizer = AutoTokenizer.from_pretrained(model_id)
tokenizer.pad_token = tokenizer.eos_token
```

```
judgment_para = """
   In Maneka Gandhi v. Union of India, (1978) 1 SCC 248, this Court held that the
 procedure established by law under Article 21 must be just, fair and reasonable.
 The principle was further explained in A.K. Gopalan v. State of Madras, AIR 1950
 SC 27, and later constitutional decisions have emphasized that fundamental rights
 are not isolated silos but must be read together.
   """
```

```
tokens = tokenizer.tokenize(judgment_para)
token_ids = tokenizer.encode(judgment_para)

print("Number of tokens:", len(token_ids))
print(tokens)
```

```
citations = [
       "Maneka Gandhi v. Union of India, (1978) 1 SCC 248",
       "A.K. Gopalan v. State of Madras, AIR 1950 SC 27",
       "(1978) 1 SCC 248",
       "AIR 1950 SC 27"
   ]

for cite in citations:
  cite_tokens = tokenizer.tokenize(cite)
  print("\nCitation:", cite)
  print("Token count:", len(cite_tokens))
  print(cite_tokens)
```
